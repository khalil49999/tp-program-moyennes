```python
### 1. Liste de toutes les matières du Semestre 1
matieres = [
    "R101", "R102", "R103", "R104", "R105", "R106", "R107", "R108", 
    "R109", "R110", "R111", "R112", "R113", "R114", "R115", 
    "SAE11", "SAE12", "SAE13", "SAE14", "SAE15", "SAE16"
]
# 2. Les coefficients officiels sont déjà enregistrés (0 quand la matière ne compte pas dans l'UE)
coeffs_ue1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coeffs_ue2 = [4, 0, 2, 8, 6, 0, 0, 2, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coeffs_ue3 = [4, 0, 2, 0, 0, 0, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

# 3. Création de la FONCTION qui calcule n'importe quelle moyenne d'UE
def calculer_moyenne(liste_notes, liste_coeffs):
    points_total = 0
    somme_coeffs = sum(liste_coeffs)
    
    for i in range(len(liste_notes)):
        points_total += liste_notes[i] * liste_coeffs[i]
        
    return points_total / somme_coeffs

# 4. On demande les notes UNE SEULE FOIS à l'étudiant
notes_etudiant = []
print("--- Saisie unique de vos notes du Semestre 1 ---")

for matiere in matieres:
    note = float(input(f"Entrez votre note pour {matiere} : "))
    notes_etudiant.append(note)

# 5. On appelle notre fonction 3 fois pour obtenir les 3 moyennes instantanément
moyenne_ue1 = calculer_moyenne(notes_etudiant, coeffs_ue1)
moyenne_ue2 = calculer_moyenne(notes_etudiant, coeffs_ue2)
moyenne_ue3 = calculer_moyenne(notes_etudiant, coeffs_ue3)

print("RÉSULTATS DU SEMESTRE 1")
print(f"Moyenne UE 1 : {moyenne_ue1:.2f}")
print(f"Moyenne UE 2 : {moyenne_ue2:.2f}")
print(f"Moyenne UE 3 : {moyenne_ue3:.2f}")
```
